```python
my_dict = {101:"Amit", 102:"Rahul",103:"Priya"}
key = int(input("Enter key to search: "))
if key in my_dict:
    print("Key exists in the dictionary")
else:
    print("Key does not exist in the dictionary")
```

```python
my_dict = {101:"Amit", 102:"Rahul",103:"Priya"}
print("Keys:")
for key in my_dict:
    print(key)
print("\nValues:")
for value in my_dict.values():
    print(value)
print("\nKeys and Values:")
for key, value in my_dict.items():
    print(key,value)
```

```python
my_dict ={}
for i in range(1,16):
    my_dict[i] = i**2
print(my_dict)
```

```python
my_dict = {"a":10, "b":20,"c":30,"d":40}
total = 0
for value in my_dict.values():
    total = total+value
print("Sum of values: ",total)
```

```python
my_dict = {101:"Amit", 102:"Rahul",103:"Priya",104 :"Shardul", 105:"Rohan"}
print("Origina; Student Info: ")
print(my_dict)
my_dict[106] = "Neha"
print("\nAfter Adding :")
print(my_dict)

del my_dict[104]
print("\nAfter Deleting :")
print(my_dict)

print("\nStudent Information :")
for key,value in my_dict.items():
    print("Roll No: ",key,"Name:",value)
```
