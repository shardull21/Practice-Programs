```python
numbers = [(10,20,40),(40,50,60),(70,80,90)]
new_list = []
for tup in numbers:
    new_tup = tup[:-1]+(100,)
    new_list.append(new_tup)
print("Original List:", numbers)
print("New List: ",new_list)
```

```python
numbers=(25,10,45,5,60,30)
maximum = max(numbers)
minimum = min(numbers)
print("Tuple :",numbers)
print("Maximum element:", maximum)
print("Minimum element:", minimum)
```
